```python

# ============================================================
# COMPLETE EXPLORATORY DATA ANALYSIS (EDA) - EXAMS.CSV
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# 2. LOAD DATASET
# ============================================================

# Upload CSV file if it is not already available
try:
    df = pd.read_csv("exams.csv")
except FileNotFoundError:
    from google.colab import files
    uploaded = files.upload()
    filename = next(iter(uploaded))
    df = pd.read_csv(filename)

print("DATASET LOADED SUCCESSFULLY")

# ============================================================
# 3. BASIC DATASET INSPECTION
# ============================================================

print("\n--- FIRST 5 ROWS ---")
display(df.head())

print("\n--- LAST 5 ROWS ---")
display(df.tail())

print("\n--- RANDOM 5 ROWS ---")
display(df.sample(min(5, len(df)), random_state=42))

print("\n--- DATASET SHAPE ---")
print(df.shape)

print("\n--- NUMBER OF ROWS ---")
print(df.shape[0])

print("\n--- NUMBER OF COLUMNS ---")
print(df.shape[1])

print("\n--- COLUMN NAMES ---")
print(df.columns.tolist())

print("\n--- DATA TYPES ---")
print(df.dtypes)

print("\n--- DATASET INFORMATION ---")
df.info()

print("\n--- NUMERICAL STATISTICS ---")
display(df.describe())

print("\n--- ALL COLUMNS SUMMARY ---")
display(df.describe(include="all"))

print("\n--- DATASET SIZE ---")
print(df.size)

print("\n--- DATASET INDEX ---")
print(df.index)

# ============================================================
# 4. NUMERICAL STATISTICAL OPERATIONS
# ============================================================

score_columns = ["math score", "reading score", "writing score"]

print("\n--- MEAN ---")
display(df[score_columns].mean())

print("\n--- MEDIAN ---")
display(df[score_columns].median())

print("\n--- MODE ---")
display(df[score_columns].mode())

print("\n--- VARIANCE ---")
display(df[score_columns].var())

print("\n--- STANDARD DEVIATION ---")
display(df[score_columns].std())

print("\n--- MINIMUM VALUES ---")
display(df[score_columns].min())

print("\n--- MAXIMUM VALUES ---")
display(df[score_columns].max())

print("\n--- COUNT OF NON-MISSING VALUES ---")
display(df.count())

print("\n--- SUM OF SCORES ---")
display(df[score_columns].sum())

print("\n--- QUARTILES ---")
display(df[score_columns].quantile([0.25, 0.50, 0.75]))

print("\n--- SCORE RANGES ---")
display(df[score_columns].agg(["min", "max", "mean", "std"]))

# ============================================================
# 5. UNIQUE VALUES AND FREQUENCY COUNTS
# ============================================================

print("\n--- UNIQUE GENDER VALUES ---")
print(df["gender"].unique())

print("\n--- NUMBER OF UNIQUE VALUES PER COLUMN ---")
display(df.nunique())

categorical_columns = [
    "gender",
    "race/ethnicity",
    "parental level of education",
    "lunch",
    "test preparation course"
]

for col in categorical_columns:
    print(f"\n--- VALUE COUNTS: {col} ---")
    display(df[col].value_counts())

    print(f"\n--- PERCENTAGES: {col} ---")
    display(df[col].value_counts(normalize=True).mul(100).round(2))

# ============================================================
# 6. MISSING VALUE ANALYSIS
# ============================================================

print("\n--- MISSING VALUES PER COLUMN ---")
display(df.isnull().sum())

print("\n--- TOTAL MISSING VALUES ---")
print(df.isnull().sum().sum())

print("\n--- MISSING VALUE PERCENTAGES ---")
display((df.isnull().mean() * 100).round(2))

print("\n--- NON-MISSING VALUES ---")
display(df.notnull().sum())

# Create a separate cleaned copy; preserve original df
df_clean = df.copy()

# Fill missing numerical values with column means
df_clean[score_columns] = df_clean[score_columns].fillna(
    df_clean[score_columns].mean()
)

# Fill missing categorical values with the most frequent category
for col in categorical_columns:
    if df_clean[col].isnull().any() and not df_clean[col].mode().empty:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("\n--- MISSING VALUES AFTER FILLING IN CLEANED COPY ---")
display(df_clean.isnull().sum())

# ============================================================
# 7. DUPLICATE VALUE ANALYSIS
# ============================================================

print("\n--- DUPLICATE ROW COUNT ---")
print(df_clean.duplicated().sum())

print("\n--- DUPLICATE ROWS ---")
display(df_clean[df_clean.duplicated()])

# Remove duplicate rows in a separate DataFrame
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\n--- SHAPE AFTER REMOVING DUPLICATES ---")
print(df_clean.shape)

# ============================================================
# 8. SORTING OPERATIONS
# ============================================================

print("\n--- SORT BY MATH SCORE ASCENDING ---")
display(df_clean.sort_values("math score").head())

print("\n--- SORT BY MATH SCORE DESCENDING ---")
display(df_clean.sort_values("math score", ascending=False).head())

print("\n--- SORT BY ALL THREE SCORES DESCENDING ---")
display(
    df_clean.sort_values(
        by=score_columns,
        ascending=[False, False, False]
    ).head()
)

# ============================================================
# 9. FILTERING OPERATIONS
# ============================================================

print("\n--- STUDENTS WITH MATH SCORE ABOVE 80 ---")
display(df_clean[df_clean["math score"] > 80])

print("\n--- STUDENTS WITH MATH SCORE BELOW 40 ---")
display(df_clean[df_clean["math score"] < 40])

print("\n--- STUDENTS WITH READING SCORE AT LEAST 90 ---")
display(df_clean[df_clean["reading score"] >= 90])

print("\n--- STUDENTS SCORING ABOVE 80 IN ALL SUBJECTS ---")
display(df_clean[(df_clean[score_columns] > 80).all(axis=1)])

print("\n--- STUDENTS WHO COMPLETED TEST PREPARATION ---")
display(
    df_clean[df_clean["test preparation course"] == "completed"]
)

print("\n--- STUDENTS WITH STANDARD LUNCH ---")
display(df_clean[df_clean["lunch"] == "standard"])

print("\n--- SELECTED COLUMNS ---")
display(
    df_clean[
        ["gender", "math score", "reading score", "writing score"]
    ].head()
)

# ============================================================
# 10. GROUPING AND AGGREGATION
# ============================================================

print("\n--- AVERAGE SCORES BY GENDER ---")
display(df_clean.groupby("gender")[score_columns].mean())

print("\n--- AVERAGE SCORES BY RACE/ETHNICITY ---")
display(df_clean.groupby("race/ethnicity")[score_columns].mean())

print("\n--- AVERAGE SCORES BY PARENTAL EDUCATION ---")
display(
    df_clean.groupby("parental level of education")[score_columns].mean()
)

print("\n--- AVERAGE SCORES BY LUNCH TYPE ---")
display(df_clean.groupby("lunch")[score_columns].mean())

print("\n--- AVERAGE SCORES BY TEST PREPARATION ---")
display(
    df_clean.groupby("test preparation course")[score_columns].mean()
)

print("\n--- MATH SCORE AGGREGATIONS BY GENDER ---")
display(
    df_clean.groupby("gender")["math score"].agg(
        ["count", "mean", "median", "min", "max", "std"]
    )
)

print("\n--- GENDER AND TEST PREPARATION GROUPING ---")
display(
    df_clean.groupby(["gender", "test preparation course"])[score_columns]
    .mean()
)

# ============================================================
# 11. CORRELATION AND COVARIANCE
# ============================================================

print("\n--- CORRELATION MATRIX ---")
display(df_clean[score_columns].corr())

print("\n--- CORRELATION WITH MATH SCORE ---")
display(df_clean[score_columns].corr()["math score"])

print("\n--- COVARIANCE MATRIX ---")
display(df_clean[score_columns].cov())

print("\n--- MATH VS READING CORRELATION ---")
print(df_clean["math score"].corr(df_clean["reading score"]))

print("\n--- READING VS WRITING CORRELATION ---")
print(df_clean["reading score"].corr(df_clean["writing score"]))

# ============================================================
# 12. FEATURE ENGINEERING
# ============================================================

# Work on a copy so the original dataset stays unchanged
df_eda = df_clean.copy()

df_eda["average score"] = df_eda[score_columns].mean(axis=1)
df_eda["total score"] = df_eda[score_columns].sum(axis=1)

print("\n--- DATASET WITH AVERAGE AND TOTAL SCORES ---")
display(df_eda.head())

print("\n--- STUDENT WITH HIGHEST AVERAGE SCORE ---")
display(df_eda.loc[[df_eda["average score"].idxmax()]])

print("\n--- STUDENT WITH LOWEST AVERAGE SCORE ---")
display(df_eda.loc[[df_eda["average score"].idxmin()]])

print("\n--- STUDENTS WITH AVERAGE SCORE AT LEAST 80 ---")
display(df_eda[df_eda["average score"] >= 80])

print("\n--- AVERAGE SCORE BY GENDER ---")
display(df_eda.groupby("gender")["average score"].mean())

print("\n--- AVERAGE SCORE BY TEST PREPARATION ---")
display(
    df_eda.groupby("test preparation course")["average score"].mean()
)

# ============================================================
# 13. HISTOGRAMS
# ============================================================

df_eda[score_columns].hist(
    figsize=(12, 4),
    bins=15,
    edgecolor="black"
)

plt.suptitle("Distribution of Exam Scores")
plt.tight_layout()
plt.show()

# ============================================================
# 14. BOXPLOT - POTENTIAL OUTLIERS
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(data=df_eda[score_columns])

plt.title("Exam Score Distribution and Potential Outliers")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

# ============================================================
# 15. SCATTER PLOT - MATH VS READING
# ============================================================

plt.figure(figsize=(7, 5))

sns.scatterplot(
    data=df_eda,
    x="math score",
    y="reading score",
    hue="gender"
)

plt.title("Math Score vs Reading Score")
plt.tight_layout()
plt.show()

# ============================================================
# 16. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(7, 5))

sns.heatmap(
    df_eda[score_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Exam Scores")
plt.tight_layout()
plt.show()

# ============================================================
# 17. COUNT PLOTS FOR CATEGORICAL COLUMNS
# ============================================================

for col in categorical_columns:
    plt.figure(figsize=(9, 4))

    sns.countplot(data=df_eda, x=col, order=df_eda[col].value_counts().index)

    plt.title(f"Student Count by {col}")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()

# ============================================================
# 18. BAR CHART - AVERAGE SCORES BY GENDER
# ============================================================

avg_scores = df_eda.groupby("gender")[score_columns].mean()

avg_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Exam Scores by Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()

# ============================================================
# 19. BAR CHART - TEST PREPARATION VS AVERAGE SCORE
# ============================================================

prep_scores = df_eda.groupby("test preparation course")[
    "average score"
].mean()

prep_scores.plot(kind="bar", figsize=(7, 4))

plt.title("Test Preparation vs Average Score")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# 20. FINAL EDA SUMMARY
# ============================================================

print("\n========== FINAL EDA SUMMARY ==========")

print("\n--- DATASET SHAPE ---")
print(df_eda.shape)

print("\n--- COLUMN NAMES ---")
print(df_eda.columns.tolist())

print("\n--- DATA TYPES ---")
print(df_eda.dtypes)

print("\n--- STATISTICAL SUMMARY ---")
display(df_eda.describe())

print("\n--- MISSING VALUES ---")
display(df_eda.isnull().sum())

print("\n--- DUPLICATE ROWS ---")
print(df_eda.duplicated().sum())

print("\n--- UNIQUE VALUES PER COLUMN ---")
display(df_eda.nunique())

print("\n--- CORRELATION MATRIX ---")
display(df_eda[score_columns].corr())

print("\n--- AVERAGE SCORES BY GENDER ---")
display(df_eda.groupby("gender")[score_columns].mean())

print("\n--- AVERAGE SCORES BY TEST PREPARATION ---")
display(
    df_eda.groupby("test preparation course")[score_columns].mean()
)

print("\n========== EDA COMPLETED SUCCESSFULLY ==========")

```
