```python

# K-MEANS CLUSTERING ON CATS VS DOGS DATASET

# Step 1: Import libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 2: Set dataset paths
train_path = r"C:\Users\user\Downloads\cats_dogs_light\cats_dogs_light\train"
test_path = r"C:\Users\user\Downloads\cats_dogs_light\cats_dogs_light\test"

# Step 3: Load and process images
def load_images(folder):

    images = []
    labels = []
    filenames = []

    for filename in os.listdir(folder):

        if filename.lower().endswith(('.jpg', '.jpeg', '.png')):

            image_path = os.path.join(folder, filename)

            try:
                img = Image.open(image_path).convert('L')
                img = img.resize((32, 32))

                # Convert image into numerical array
                img_array = np.array(img) / 255.0

                images.append(img_array.flatten())

                # Extract actual label from filename
                if filename.lower().startswith('cat'):
                    labels.append('Cat')
                elif filename.lower().startswith('dog'):
                    labels.append('Dog')
                else:
                    labels.append('Unknown')

                filenames.append(filename)

            except Exception as e:
                print("Error loading:", filename, e)

    return np.array(images), np.array(labels), filenames


# Load training images
X, y, filenames = load_images(train_path)

print("Dataset shape:", X.shape)
print("Total images:", len(X))

# Step 4: Display dataset information
print("\nNumber of Cats:", np.sum(y == 'Cat'))
print("Number of Dogs:", np.sum(y == 'Dog'))

# Step 5: Apply PCA for dimensionality reduction
pca = PCA(n_components=50, random_state=42)
X_pca = pca.fit_transform(X)

print("\nReduced dataset shape:", X_pca.shape)

print("Variance retained:",
      round(sum(pca.explained_variance_ratio_) * 100, 2), "%")

# Step 6: Standardize the features
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_pca)

# Step 7: Apply Elbow Method
inertia = []

for k in range(1, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)
    inertia.append(model.inertia_)

# Step 8: Plot Elbow graph
plt.figure(figsize=(8, 5))

plt.plot(range(1, 11), inertia, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method - Cats vs Dogs Dataset")

plt.grid(True)
plt.show()

# Step 9: Apply K-Means with K = 2
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

# Step 10: Display cluster assignments
print("\nFirst 20 Cluster Assignments:")

for i in range(min(20, len(clusters))):

    print(
        filenames[i],
        "Actual:", y[i],
        "Cluster:", clusters[i]
    )

# Step 11: Visualize clusters using PCA
pca_visual = PCA(n_components=2, random_state=42)

X_visual = pca_visual.fit_transform(X_scaled)

plt.figure(figsize=(8, 5))

plt.scatter(
    X_visual[:, 0],
    X_visual[:, 1],
    c=clusters,
    cmap='viridis',
    s=40
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.title("K-Means Clustering - Cats vs Dogs (K=2)")

plt.colorbar(label="Cluster")
plt.show()

# Step 12: Display cluster distribution
print("\nCluster Distribution:")

print(pd.Series(clusters).value_counts().sort_index())

# Step 13: Compare clusters with actual labels
results = pd.DataFrame({
    "Filename": filenames,
    "Actual Label": y,
    "Cluster": clusters
})

print("\nActual Labels vs Predicted Clusters:")

print(pd.crosstab(
    results["Actual Label"],
    results["Cluster"]
))

# Step 14: Save results
results.to_csv("cats_dogs_kmeans_results.csv", index=False)

print("\nK-Means Clustering completed successfully!")

print("Results saved as cats_dogs_kmeans_results.csv")
```
