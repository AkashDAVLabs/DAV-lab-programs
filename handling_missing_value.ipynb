```python

# ============================================================
# HANDLING MISSING VALUES AND ENCODING CATEGORICAL VARIABLES
# Dataset: retail_sales_dataset.csv
# ============================================================

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from google.colab import files

# STEP 1: Upload the dataset
uploaded = files.upload()
file_name = next(iter(uploaded))

# STEP 2: Read the dataset
df = pd.read_csv(file_name)

print("ORIGINAL DATASET")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

# STEP 3: Identify missing values
print("\nMissing Values in Each Cell:")
print(df.isnull())

print("\nTotal Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values in Dataset:")
print(df.isnull().sum().sum())

# STEP 4: Mean Imputation
# Fill missing Age values with the mean Age
df['Age'] = df['Age'].fillna(df['Age'].mean())

print("\nMean Age:", df['Age'].mean())
print("\nDataset After Mean Imputation:")
print(df.head())

# STEP 5: Median Imputation
# Fill missing Price per Unit values with the median
df['Price per Unit'] = df['Price per Unit'].fillna(
    df['Price per Unit'].median()
)

print("\nMedian Price per Unit:", df['Price per Unit'].median())
print("\nDataset After Median Imputation:")
print(df.head())

# STEP 6: Forward Fill
# Fill missing Product Category values
# using the previous available value
print("\nBefore Forward Fill:")
print(df['Product Category'].head(10))

df['Product Category'] = df['Product Category'].ffill()

print("\nAfter Forward Fill:")
print(df['Product Category'].head(10))

# STEP 7: Label Encoding
# Create a separate copy so the original Gender
# column remains available
label_df = df.copy()

encoder = LabelEncoder()

label_df['Gender_Encoded'] = encoder.fit_transform(
    label_df['Gender'].astype(str)
)

print("\nLabel Encoding Results:")
print(label_df[['Gender', 'Gender_Encoded']].head(10))

print("\nGender Classes:")
print(encoder.classes_)

# STEP 8: One-Hot Encoding
# Create a separate column for each product category
encoded_df = pd.get_dummies(
    df,
    columns=['Product Category'],
    dtype=int
)

print("\nOne-Hot Encoding Results:")
print(encoded_df.head())

print("\nOne-Hot Encoded Column Names:")
print([
    col for col in encoded_df.columns
    if col.startswith('Product Category_')
])

# STEP 9: Verify missing values after processing
print("\nMissing Values After Processing:")
print(encoded_df.isnull().sum())

# STEP 10: Save the processed dataset
output_file = 'retail_sales_dataset_processed.csv'
encoded_df.to_csv(output_file, index=False)

print("\nProcessed dataset saved successfully!")

# STEP 11: Download the processed dataset
files.download(output_file)
```
